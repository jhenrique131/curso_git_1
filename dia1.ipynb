```python
import pandas as pd
import io

# 1. Carregar a base de dados fornecida
df = pd.read_csv("/content/drive/MyDrive/Ciencia de Dados - SEBRAE/ginasio_checkins.csv") # Substitua pelo caminho do seu arquivo

total_inicial = len(df)

# --- ETAPA 1: Tratamento de Duplicatas ---
# Identificar o número total de linhas duplicadas na base inteira
total_duplicadas = df.duplicated().sum()

# Remover mantendo o primeiro registro ('first') pois o app salvou exatamente o mesmo dado
df_no_dup = df.drop_duplicates(keep='first')
total_apos_dup = len(df_no_dup)
linhas_removidas_dup = total_inicial - total_apos_dup

# --- ETAPA 2: Remoção de Outliers (Duração Fora do Padrão) ---
# Calcular os quartis e o IQR para a coluna duracao_min
Q1 = df_no_dup['duracao_min'].quantile(0.25)
Q3 = df_no_dup['duracao_min'].quantile(0.75)
IQR = Q3 - Q1

# Definir os limites estatísticos
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# Filtrar a base mantendo apenas os registros dentro dos limites
df_clean = df_no_dup[(df_no_dup['duracao_min'] >= limite_inferior) &
                      (df_no_dup['duracao_min'] <= limite_superior)].copy()

total_apos_outliers = len(df_clean)
linhas_removidas_outliers = total_apos_dup - total_apos_outliers

# --- ETAPA 3: Padronização do Desempenho (Z-Score) ---
# Aplicar o Z-score de forma agrupada para cada unidade
df_clean['z_score_desempenho'] = df_clean.groupby('unidade')['pontuacao_desempenho'].transform(
    lambda x: (x - x.mean()) / x.std()
)

# --- EXIBIÇÃO DOS RESULTADOS ---
print("=== RELATÓRIO DE PROCESSAMENTO EM ETAPAS ===")
print(f"0. Registros Iniciais: {total_inicial} linhas.")
print(f"1. Remoção de Duplicatas: {linhas_removidas_dup} registros eliminados (Total restante: {total_apos_dup}).")
print(f"2. Remoção de Outliers (IQR): {linhas_removidas_outliers} registros de treinos absurdos eliminados (Total final: {total_apos_outliers}).")
print(f"   * Limite superior aceitável de treino: {limite_superior:.1f} minutos.")

print("\n=== AMOSTRA DOS DADOS TRATADOS E PADRONIZADOS ===")
print(df_clean[['id_checkin', 'unidade', 'duracao_min', 'pontuacao_desempenho', 'z_score_desempenho']].head())
```

```python
"""
==========================================================================
 MINI-PROJETO AVALIATIVO - MÓDULO 1
 Análise Exploratória de Dados (AED) - Base Varejo
 Turma: Analise_de_Dados_T6
==========================================================================

OBJETIVO
--------
Realizar uma Análise Exploratória de Dados (AED) sobre a base "Varejo",
que contém registros reais de compras (um item por linha). O script:

  1) Carrega os dados e descreve sua estrutura;
  2) Identifica problemas de qualidade (nulos, duplicatas, inconsistências);
  3) Realiza a limpeza mínima necessária (nulos, duplicatas, tipos);
  4) Valida e explora a lógica de agrupamento por compra (CO_ID);
  5) Calcula estatísticas descritivas da coluna de número de filhos
     do cliente (CL_FHL);
  6) Explora padrões via groupby()/pivot_table();
  7) Imprime um relatório final com as principais conclusões.

COMO EXECUTAR
--------------
  - VSCode  : abra este arquivo e rode "python Miniprojeto_Varejo_AED.py"
              (o arquivo Base_Varejo.csv deve estar na mesma pasta).
  - Colab   : faça upload do Base_Varejo.csv para o ambiente e rode
              todas as células (ou execute o arquivo .py diretamente).

DICIONÁRIO DE DADOS (colunas relevantes)
-----------------------------------------
  DATA        -> data da compra (dd/mm/aaaa)
  CO_ID       -> identificador da COMPRA (se repete para os itens da
                 mesma compra; NÃO é um identificador de linha único)
  CL_ID       -> identificador do cliente
  CL_GENERO   -> gênero do cliente (M/F)
  CL_EC       -> estado civil do cliente (código)
  CL_FHL      -> número de filhos do cliente
  CL_SEG      -> segmento do cliente (A/B/C)
  PR_ID       -> identificador do produto
  PR_CAT      -> categoria do produto (contém "#N/D" quando ausente)
  PR_NOME     -> nome do produto
==========================================================================
"""

import pandas as pd
import numpy as np

pd.set_option("display.width", 120)
pd.set_option("display.max_columns", 20)

ARQUIVO_ENTRADA = "/content/drive/MyDrive/Ciencia de Dados - SEBRAE/TrabalhoPython/Base_Varejo.csv"
ARQUIVO_SAIDA = "/content/drive/MyDrive/Ciencia de Dados - SEBRAE/TrabalhoPython/df_limpo.csv"
SEPARADOR = ";"
CATEGORIA_AUSENTE = "#N/D"


def linha(titulo=""):
    """Imprime um separador visual no relatório do terminal."""
    print("\n" + "=" * 78)
    if titulo:
        print(titulo)
        print("=" * 78)
```

```python
# ==========================================================================
# ETAPA 1 - CARGA DOS DADOS
# ==========================================================================
def carregar_dados(caminho: str) -> pd.DataFrame:
    """Carrega o CSV com pandas e exibe uma visão geral da base."""
    df = pd.read_csv(caminho, sep=SEPARADOR)

    linha("ETAPA 1 - CARGA E VISÃO GERAL DOS DADOS")
    print(f"Arquivo carregado: {caminho}")
    print(f"Número de registros (linhas): {df.shape[0]}")
    print(f"Número de colunas: {df.shape[1]}")
    print("\nColunas e tipos de dados originais:")
    print(df.dtypes)
    print("\nAmostra das 5 primeiras linhas:")
    print(df.head())

    return df


# ==========================================================================
# ETAPA 2 - VERIFICAÇÃO DE PROBLEMAS (qualidade dos dados)
# ==========================================================================
def diagnosticar_problemas(df: pd.DataFrame) -> None:
    """Reporta nulos, duplicatas e inconsistências conhecidas da base."""
    linha("ETAPA 2 - DIAGNÓSTICO DE QUALIDADE DOS DADOS")

    # --- 2.1 Valores nulos por coluna ---
    print("Valores nulos por coluna:")
    print(df.isnull().sum())

    # A leitura do CSV com ';' costuma gerar colunas fantasmas "Unnamed: N"
    # por causa de ';' sobrando no fim de cada linha do arquivo original.
    colunas_fantasma = [c for c in df.columns if c.startswith("Unnamed")]
    if colunas_fantasma:
        print(f"\nColunas 'fantasma' (100% nulas, geradas por ';' extra "
              f"no CSV): {colunas_fantasma}")

    # --- 2.2 Duplicatas ---
    duplicadas_completas = df.duplicated().sum()
    print(f"\nLinhas totalmente duplicadas: {duplicadas_completas}")

    # Duplicata "de negócio": o mesmo produto (PR_ID) aparecendo mais de uma
    # vez dentro da MESMA compra (CO_ID) - indica erro de digitação/leitura
    # de código de barras no PDV, e não uma segunda unidade comprada.
    dup_item_na_compra = df.duplicated(subset=["CO_ID", "PR_ID"]).sum()
    print(f"Linhas com o mesmo PR_ID repetido dentro da mesma compra "
          f"(CO_ID): {dup_item_na_compra}")

    # --- 2.3 Inconsistências conhecidas ---
    # (a) Categoria de produto ausente, representada pela string "#N/D"
    qtd_cat_ausente = (df["PR_CAT"] == CATEGORIA_AUSENTE).sum()
    pct_cat_ausente = qtd_cat_ausente / len(df) * 100
    print(f"\nRegistros com categoria de produto ausente "
          f"('{CATEGORIA_AUSENTE}'): {qtd_cat_ausente} "
          f"({pct_cat_ausente:.2f}% da base)")

    # (b) Datas inválidas (fora do formato dd/mm/aaaa ou não conversíveis)
    datas_convertidas = pd.to_datetime(df["DATA"], format="%d/%m/%Y",
                                        errors="coerce")
    qtd_datas_invalidas = datas_convertidas.isna().sum()
    print(f"Registros com data inválida/não reconhecida: "
          f"{qtd_datas_invalidas}")

    # (c) Verifica se CL_GENERO e CL_SEG possuem apenas valores esperados
    print(f"\nValores únicos em CL_GENERO: {sorted(df['CL_GENERO'].unique())}")
    print(f"Valores únicos em CL_SEG: {sorted(df['CL_SEG'].unique())}")



```

```python
# ==========================================================================
# ETAPA 3 - LIMPEZA MÍNIMA DOS DADOS
# ==========================================================================
def limpar_dados(df: pd.DataFrame) -> pd.DataFrame:
    """
    Aplica as três limpezas mínimas exigidas:
      (1) tratar nulos;
      (2) eliminar duplicatas relevantes;
      (3) ajustar tipos de dados (DATA -> datetime).
    Também trata a inconsistência da categoria "#N/D".
    """
    linha("ETAPA 3 - LIMPEZA DOS DADOS")
    df = df.copy()
    registros_iniciais = len(df)

    # --- (1) Tratamento de nulos --------------------------------------
    # As colunas "Unnamed: *" são 100% nulas e não trazem nenhuma
    # informação (são artefato de leitura do CSV) -> removê-las é seguro
    # e não descarta nenhum dado real.
    colunas_fantasma = [c for c in df.columns if c.startswith("Unnamed")]
    if colunas_fantasma:
        df = df.drop(columns=colunas_fantasma)
        print(f"Colunas 100% nulas removidas: {colunas_fantasma}")

    # Nas colunas de negócio não há valores NaN reais nesta base; o único
    # "vazio" é a string "#N/D" em PR_CAT. Optamos por IMPUTAR (não
    # remover a linha), trocando "#N/D" por "NAO INFORMADO":
    #   - a compra em si é um dado válido (cliente, data, produto);
    #   - descartar a linha jogaria fora informação de venda real só
    #     porque a categoria do produto não foi cadastrada;
    #   - manter uma categoria explícita ("NAO INFORMADO") permite
    #     continuar agrupando por categoria sem perder esses registros.
    qtd_antes = (df["PR_CAT"] == CATEGORIA_AUSENTE).sum()
    df["PR_CAT"] = df["PR_CAT"].replace(CATEGORIA_AUSENTE, "NAO INFORMADO")
    print(f"Categoria '{CATEGORIA_AUSENTE}' imputada como 'NAO INFORMADO' "
          f"em {qtd_antes} registros.")

    # --- (2) Remoção de duplicatas relevantes ---------------------------
    # Removemos linhas em que o mesmo PR_ID aparece repetido dentro da
    # MESMA compra (CO_ID), pois representam o mesmo item lançado mais de
    # uma vez por erro de leitura/digitação no ponto de venda, e não uma
    # segunda unidade realmente comprada.
    antes = len(df)
    df = df.drop_duplicates(subset=["CO_ID", "PR_ID"], keep="first")
    removidas = antes - len(df)
    print(f"Duplicatas (mesmo PR_ID na mesma compra CO_ID) removidas: "
          f"{removidas}")

    # --- (3) Ajuste de tipos de dados -----------------------------------
    # Converte a coluna DATA (texto dd/mm/aaaa) para datetime real,
    # habilitando análises temporais (ano, mês, dia da semana etc.).
    df["DATA"] = pd.to_datetime(df["DATA"], format="%d/%m/%Y",
                                 errors="coerce")

    # Colunas categóricas/textuais -> tipo 'category' (mais eficiente em
    # memória e deixa explícito que são variáveis categóricas).
    for col in ["CL_GENERO", "CL_SEG", "PR_CAT"]:
        df[col] = df[col].astype("category")

    # Colunas que são, de fato, identificadores/códigos inteiros.
    for col in ["CO_ID", "CL_ID", "CL_EC", "CL_FHL", "PR_ID"]:
        df[col] = df[col].astype("int64")

    print("\nTipos de dados após a limpeza:")
    print(df.dtypes)

    registros_finais = len(df)
    print(f"\nRegistros antes da limpeza: {registros_iniciais}")
    print(f"Registros após a limpeza:  {registros_finais}")
    print(f"Total de linhas removidas: "
          f"{registros_iniciais - registros_finais}")

    return df


```

```python
# ==========================================================================
# ETAPA 4 - VALIDAÇÃO DA REGRA CO_ID (item x compra)
# ==========================================================================
def validar_regra_co_id(df: pd.DataFrame) -> pd.DataFrame:
    """
    Confirma que CO_ID identifica a COMPRA (não a linha) e constrói um
    resumo por compra (número de itens, categorias distintas, etc.).
    """
    linha("ETAPA 4 - VALIDAÇÃO DO IDENTIFICADOR DE COMPRA (CO_ID)")

    total_linhas = len(df)
    total_compras = df["CO_ID"].nunique()
    print(f"Total de linhas (itens comprados): {total_linhas}")
    print(f"Total de compras distintas (CO_ID únicos): {total_compras}")
    print(f"Média de itens por compra: {total_linhas / total_compras:.2f}")

    # Resumo por compra: quantos itens, quantas categorias diferentes,
    # cliente e data da compra.
    resumo_compras = df.groupby("CO_ID").agg(
        cliente=("CL_ID", "first"),
        data_compra=("DATA", "first"),
        qtd_itens=("PR_ID", "count"),
        qtd_categorias_distintas=("PR_CAT", "nunique"),
    ).reset_index()

    print("\nDistribuição de itens por compra (describe):")
    print(resumo_compras["qtd_itens"].describe())

    print("\nExemplo: primeiras 5 compras resumidas:")
    print(resumo_compras.head())

    return resumo_compras


```

```python
# ==========================================================================
# ETAPA 5 - ESTATÍSTICA DESCRITIVA (número de filhos do cliente)
# ==========================================================================
def estatisticas_filhos(df: pd.DataFrame) -> None:
    """
    Calcula estatísticas descritivas da coluna CL_FHL (número de filhos
    do cliente). Como cada cliente se repete em várias linhas (uma por
    item comprado), a estatística é calculada sobre a base de CLIENTES
    ÚNICOS, para não pesar o resultado a favor de quem comprou mais itens.
    """
    linha("ETAPA 5 - ESTATÍSTICA DESCRITIVA: Nº DE FILHOS DO CLIENTE (CL_FHL)")

    clientes_unicos = df.drop_duplicates(subset="CL_ID")
    filhos = clientes_unicos["CL_FHL"]

    print(f"Base considerada: {len(clientes_unicos)} clientes únicos "
          f"(evita contar o mesmo cliente várias vezes por item comprado).")

    print(f"\nContagem (n)     : {filhos.count()}")
    print(f"Média            : {filhos.mean():.3f}")
    print(f"Mediana          : {filhos.median():.3f}")
    print(f"Desvio padrão    : {filhos.std():.3f}")
    print(f"Moda             : {filhos.mode().tolist()}")
    print(f"Mínimo           : {filhos.min()}")
    print(f"Máximo           : {filhos.max()}")
    print(f"1º quartil (25%) : {filhos.quantile(0.25)}")
    print(f"2º quartil (50%) : {filhos.quantile(0.50)}")
    print(f"3º quartil (75%) : {filhos.quantile(0.75)}")

    print("\nResumo via describe():")
    print(filhos.describe())

    print("\nDistribuição de frequência (nº de filhos -> quantidade de "
          "clientes):")
    print(filhos.value_counts().sort_index())


```

```python
# ==========================================================================
# ETAPA 6 - PADRÕES DE AGRUPAMENTO
# ==========================================================================
def explorar_agrupamentos(df: pd.DataFrame) -> None:
    """Explora ao menos dois agrupamentos relevantes para o negócio."""
    linha("ETAPA 6 - PADRÕES DE AGRUPAMENTO")

    # --- Agrupamento 1: itens vendidos e clientes distintos por gênero ---
    print("[Agrupamento 1] Itens vendidos e clientes distintos por gênero:")
    por_genero = df.groupby("CL_GENERO", observed=True).agg(
        qtd_itens_vendidos=("PR_ID", "count"),
        clientes_distintos=("CL_ID", "nunique"),
        compras_distintas=("CO_ID", "nunique"),
    )
    print(por_genero)

    # --- Agrupamento 2: itens vendidos por categoria de produto ---------
    print("\n[Agrupamento 2] Itens vendidos por categoria de produto "
          "(top categorias):")
    por_categoria = (
        df.groupby("PR_CAT", observed=True)
        .size()
        .sort_values(ascending=False)
        .rename("qtd_itens_vendidos")
    )
    print(por_categoria)

    # --- Agrupamento 3 (extra): pivot_table gênero x categoria ----------
    print("\n[Agrupamento 3] Tabela dinâmica: qtd. de itens por "
          "categoria x segmento de cliente (CL_SEG):")
    tabela_dinamica = pd.pivot_table(
        df, index="PR_CAT", columns="CL_SEG", values="PR_ID",
        aggfunc="count", fill_value=0, observed=True,
    )
    print(tabela_dinamica)

    # --- Agrupamento 4 (extra): vendas ao longo do tempo (por mês) ------
    print("\n[Agrupamento 4] Quantidade de itens vendidos por mês:")
    vendas_por_mes = (
        df.assign(ANO_MES=df["DATA"].dt.to_period("M"))
        .groupby("ANO_MES", observed=True)
        .size()
    )
    print(vendas_por_mes)


```

```python
# ==========================================================================
# ETAPA 7 - CONCLUSÕES E RELATÓRIO FINAL
# ==========================================================================
def gerar_conclusoes(df_bruto: pd.DataFrame, df_limpo: pd.DataFrame) -> None:
    """Imprime um bloco de conclusões com base nos resultados obtidos."""
    linha("ETAPA 7 - CONCLUSÕES")

    total_bruto = len(df_bruto)
    total_limpo = len(df_limpo)
    total_compras = df_limpo["CO_ID"].nunique()
    total_clientes = df_limpo["CL_ID"].nunique()
    genero_top = df_limpo["CL_GENERO"].value_counts().idxmax()
    categoria_top = df_limpo["PR_CAT"].value_counts().idxmax()
    media_itens_compra = total_limpo / total_compras

    conclusoes = [
        f"A base original possuía {total_bruto} linhas (itens comprados); "
        f"após a limpeza (remoção de duplicatas de item por compra e de "
        f"colunas 100% vazias), restaram {total_limpo} linhas válidas.",

        f"A base representa {total_clientes} clientes únicos realizando "
        f"{total_compras} compras distintas, com uma média de "
        f"{media_itens_compra:.1f} itens por compra (CO_ID).",

        f"O gênero com maior volume de itens comprados é '{genero_top}', "
        f"o que pode orientar campanhas e sortimento direcionados.",

        f"A categoria de produto mais vendida é '{categoria_top}', "
        f"indicando o principal motor de vendas da rede.",

        "A coluna de número de filhos do cliente (CL_FHL) apresenta baixa "
        "variabilidade (poucos valores distintos, de 0 a 4), o que é "
        "coerente com um dado cadastral discreto e sugere que ela é mais "
        "útil como variável de segmentação do que de tendência central.",

        "Problemas remanescentes: (a) a base não possui coluna de valor "
        "monetário (preço/quantidade), o que limita análises de "
        "faturamento e ticket médio; (b) mesmo após a imputação, a "
        "categoria 'NAO INFORMADO' (antigo '#N/D') ainda representa "
        "produtos sem classificação de categoria, o que pode ser "
        "melhorado com um cadastro de produtos mais completo; (c) não há "
        "verificação cruzada de que os dados cadastrais do cliente "
        "(gênero, estado civil, filhos) são estáveis ao longo do tempo "
        "para o mesmo CL_ID.",
    ]

    for i, c in enumerate(conclusoes, start=1):
        print(f"{i}. {c}\n")


```

```python
# ==========================================================================
# EXECUÇÃO PRINCIPAL
# ==========================================================================
def main():
    df_bruto = carregar_dados(ARQUIVO_ENTRADA)
    diagnosticar_problemas(df_bruto)
    df_limpo = limpar_dados(df_bruto)
    validar_regra_co_id(df_limpo)
    estatisticas_filhos(df_limpo)
    explorar_agrupamentos(df_limpo)
    gerar_conclusoes(df_bruto, df_limpo)

    # Salva a base limpa para uso em análises/dashboards futuros.
    df_limpo.to_csv(ARQUIVO_SAIDA, sep=";", index=False)
    linha("FIM DA EXECUÇÃO")
    print(f"Base limpa exportada para: {ARQUIVO_SAIDA}")


if __name__ == "__main__":
    main()
```
